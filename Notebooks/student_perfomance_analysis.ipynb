```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Loading the dataset
df=pd.read_csv('../Data/raw_student_performance_500.csv')
print(df.to_string())
print('=========================================================================================')
# Inspecting the dataset
# First five columns
print(df.head())
print('=========================================================================================')
# Last 5 columns
print(df.tail())
print('=========================================================================================')
# Checking the number of columns
print(df.columns)
print('=========================================================================================')
# Checking the shape of the dataset
print(df.shape)
print('=========================================================================================')
# Checking the dataset information
print(df.info())
print('=========================================================================================')
# Checking the dataset description
print(df.describe())
```

```python
# Checking for missing values per column
print(df.isnull().sum())
print('=========================================================================================')
# Calculating the percentage of missing values per column
percent=(df.isnull().mean()*100).round(2)
print(percent)

print('=========================================================================================')
```

```python
# Handling missing values
# 1. Age Column
df['Age']=df['Age'].fillna(
    round(df['Age'].mean(),0)
)
print(df.to_string())
```

```python
# 2. Study Hours column
df['Study_Hours']=df['Study_Hours'].fillna(
    round(df['Study_Hours'].mean(),2)
)
print(df.to_string())
```

```python
# 3. Attendance Column
df['Attendance']=df['Attendance'].fillna(
    round(df['Attendance'].mean(),2)
)
print(df.to_string())
```

```python
# 4. Assignment Score
df['Assignment_Score']=df['Assignment_Score'].fillna(
    round(df['Assignment_Score'].mode(),2)
)
print(df.to_string())
```

```python
# 6. Final Score
df['Final_Score']=df['Final_Score'].fillna(
    round(df['Final_Score'].mode(),2)
)

print(df.to_string())
```

```python
# Checking for the remaining missing values
print(df.isnull().sum())
```

```python
# Handling missing values in genger column
df['Gender']=df['Gender'].fillna('Unknown')
print(df.to_string())
```

```python
# Handling missing values in the name column
df['Student_Name']=df['Student_Name'].fillna('Unknown')
print(df.to_string())
```

```python
# Handling missing values in class and year of study column
df['Class']=df['Class'].fillna('Unknown')
df['Year_of_Study']=df['Year_of_Study'].fillna('Unknown')
print(df.to_string())
```

```python
# Checking for the remaining missing values
print(df.isnull().sum())
```

```python
# Checking for duplicates
print(df.duplicated().sum())
```

```python
# Replacing inconsistent values in class column
df['Class']=df['Class'].replace({
    'A':'Class A',
    'B':'Class B',
    'C':'Class C',
    'D':'Class D',
    'class a': 'Class A',
    'class b':'Class B',
    'class c':'Class C',
    'class d': 'Class D'
})
print(df.to_string())
```

```python
# DATA ANALYSIS
# Category A Understanding Academic perfomance
# Q1: What is the average final examination score
average=round(df['Final_Score'].mean(),2)
print('Average Final Score: ',average)

# Q2: What is the median final examination score?
median=df['Final_Score'].median()
print('Median Score: ',median)

# Q3: What is the most frequently occurring final score?
mode=df['Final_Score'].mode()
print('Most Frequently Occurring final Score: ',mode)

# Q4: What is the highest and lowest final score?
highest=df['Final_Score'].max()
lowest=df['Final_Score'].min()

print('Highest Score: ',highest)
print('Lowest Score: ',lowest)

# Q5: How much do final scores vary among students?
range=highest-lowest
variance=round(df['Final_Score'].var(),2)
std_dev=round(df['Final_Score'].std(),2)

print('Final Score Range: ',range)
print('Final Score Variance: ',variance)
print('Final Score Standard Deviation: ',std_dev)

# Q6: What percentage of students scored at least 70?
student_with_at_least_70=df[df['Final_Score']>=70].sum()
total_students=df['Final_Score'].count()
percentage=round(len(student_with_at_least_70)/total_students*100,2)
print('Percentage Of Students Who Scored at least 70: ',percentage,'%')

# Q7: What percentage of students passed?
# Adding a column to the dataframe called pass rate
df.loc[df['Final_Score']<50, 'Pass_Rate']='Fail'
df.loc[df['Final_Score']>50, 'Pass_Rate']='Pass'
df.loc[df['Final_Score']>70, 'Pass_Rate']='High Score'
df

# Counting the number of students who passed
passed=df[df['Final_Score']>50].sum()
total_students_passed=df['Final_Score'].count()
percentage_Pass=round(len(passed)/total_students_passed*100,2)
print('Percentage Of Student Who Passed: ',percentage_Pass,'%')

minimum=df['Final_Score'].min()
print('Minimum Final Score: ',minimum)
maximum=df['Final_Score'].max()
print('Maximum Final Scores: ',maximum)
```

```python
# Plotting a histogram to show how final score are distributed
fig,ax=plt.subplots(figsize=(12,6))
bars=df.dropna(subset=['Final_Score']).plot(y='Final_Score',
        ax=ax,
        kind='hist',
        color='b',
        bins=10,
        edgecolor='black')

# Identifying the Data Distribution
mean_score=df['Final_Score'].mean()
median_score=df['Final_Score'].median()

# Drawing Vertical lines
ax.axvline(mean_score,
           ls='-',
           c='r',
           lw=3,
           label='Mean')
ax.axvline(median_score,
           ls=':',
           c='g',
           lw=4,
           label='Median')
ax.set_title('Final Score Distribution')
ax.set_xlabel('Final Score')
ax.set_ylabel('Number Of Students')
ax.bar_label(bars.containers[0])
ax.grid(axis='y')
ax.legend()
fig.savefig('../Visualizations/Final Score Distribution.png',
            dpi=300,
            bbox_inches='tight')
plt.tight_layout()
plt.show()
```

```python
# CATEGORY B: STUDY HOURS AND ACADEMIC PERFOMANCE
# Q1 : How many hours do students study on average?
mean_hours=round(df['Study_Hours'].mean(),4)
median_hours=df['Study_Hours'].median()

print('Average Hours Of Study: ',mean_hours)
print('Median Hours Of Study: ',median_hours)
print('-------------------------------------------------------------------------')
# Q2: What is the relationship between study hours and final scores?
correlation=round(df['Study_Hours'].corr(df['Final_Score']),5)
print('Correlation between Study Hours and Final Score: ',correlation)
print('-------------------------------------------------------------------------')
# Identifying the strength of correlation
if(correlation==1.00):
    print('Perfect Positive Correlation')
elif((correlation>=0.70) & (correlation<=0.99)):
    print('Strong Positive Correlation')
elif((correlation>=0.30)& (correlation<=0.68)):
    print('Moderate Positive Correlation')
elif((correlation>=0.01)& (correlation<=0.29)):
    print('Weak Positive correlation')
elif(correlation==0.00):
    print('No Linear Correlation')
elif((correlation>=-0.01)&(correlation<=-0.29)):
    print('Weak Negative Correlation')
elif((correlation>=-0.30)& (correlation<=-0.69)):
    print('Moderate Negative correlation')
elif((correlation>=-0.70)& (correlation<=-0.99)):
    print('Strong Negative correlation')
elif(correlation==-1):
    print('Perfect Negative Correlation')
print('-------------------------------------------------------------------------') 
# Q3: Do students studying more than 6 hours have higher average scores than those studying 6 hours or less?
# Average of students studying more than 6 hours
more_than_6Hours=round(df[df['Study_Hours']>6]['Final_Score'].mean(),3)

# Average of students Studying laes than 6 Hours
less_than_6Hours=round(df[df['Study_Hours']<=6]['Final_Score'].mean(),3)

# Displyaing the results
print('Average Of Students Studying more than 6 Hours: ',more_than_6Hours)
print('-------------------------------------------------------------------------')
print('Average Of Students Studying Less than 6 Hours:  ',less_than_6Hours)

if(more_than_6Hours>less_than_6Hours):
    print('Yes! Students Studying more than 6 Hours tend to Score Higher')
elif(more_than_6Hours<less_than_6Hours):
    print('No students studying more hours have lower score')
else:
    print('Both have the same score')
print('-------------------------------------------------------------------------')
# Q4: What is the average score for students studying fewer than 3 hours?
less_than_3Hours=round(df[df['Study_Hours']<3]['Final_Score'].mean(),4)
print('Average Score Of students Studying less than 3 Hours: ',less_than_3Hours)
print('-------------------------------------------------------------------------')
# Q5: What is the average score for students studying 3–6 hours?
between_3and6=round(df[(df['Study_Hours']>=3) & (df['Study_Hours']<=6)]['Final_Score'].mean(),4)
print('Average Score For Students Studying 3 and 6 Hours: ',between_3and6)
print('-------------------------------------------------------------------------')
# Q6: What is the average score for students studying more than 6 hours?
more_than_6Hour=round(df[df['Study_Hours']>6]['Final_Score'].mean(),4)
print('Average Score for Students Studying more than 6 Hours: ',more_than_6Hour)
print('-------------------------------------------------------------------------')
# Q7: Are there students who study for many hours but still achieve low scores?
low_performers=df[(df['Study_Hours']>6) & (df['Final_Score']<50)]
print('-------------------------------------------------------------------------')
print(low_performers.to_string())

# Q8: 
```

```python
# Visual to show the relationship between study hours and final score
fig,ax=plt.subplots(figsize=(12,6))
clean_df = df.dropna(subset=['Study_Hours', 'Final_Score'])
clean_df.plot(x='Study_Hours',
        y='Final_Score',
        kind='scatter',
        alpha=0.7,
        ax=ax)

ax.set_title('Relationship Between Study Hours and Final Score')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Final Score')

# Drawing a trend line to identify the relationship
slope,intercept=np.polyfit(clean_df['Study_Hours'],
                           clean_df['Final_Score'],
                           1)
# Generate sorted x-values for a straight trend line
x_line = np.linspace(
    clean_df['Study_Hours'].min(),
    clean_df['Study_Hours'].max(),
    100
)
# Calculate the corresponding y-values
y_line = slope * x_line + intercept

# Plotting the line
ax.plot(x_line,
        y_line,
        ls='-',
        c='r',
        lw=3,
        label='Trend Line',
        zorder=3)
ax.legend()
ax.grid(axis='y')
fig.savefig('../Visualizations/Relationship Between Study Hours and Final Score.png',
            dpi=300,
            bbox_inches='tight')
plt.tight_layout()
plt.show()
print('Slope: ',slope)
print('Intercept: ',intercept)
print('Minimum Score: ',df['Study_Hours'].min())
print('Maximum Score: ',df['Study_Hours'].max())
```

```python
# CATEGORY C
# ATTENDACE AND ACADEMIC PERFORMANCE
# Q1: What is the average student attendance rate?
mean_attendance=round(df['Attendance'].mean(),4)
median_attendance=df['Attendance'].median()
print('---------------------------------------------------')
print('Mean Attendance Rate: ',mean_attendance)
print('Median Attendance Rate: ',median_attendance)
print('---------------------------------------------------')
# Q2: What is the relationship between attendance and final scores?
correlation1=round(df['Attendance'].corr(df['Final_Score']),4)
print('Correlation Between Attendance and final score: ',correlation1)
print('---------------------------------------------------')
# Identifying the strength of correlation
if(correlation1==1.00):
    print('Perfect Positive Correlation')
elif((correlation1>=0.70) & (correlation1<=0.99)):
    print('Strong Positive Correlation')
elif((correlation1>=0.30)& (correlation1<=0.68)):
    print('Moderate Positive Correlation')
elif((correlation1>=0.01)& (correlation1<=0.29)):
    print('Weak Positive correlation')
elif(correlation1==0.00):
    print('No Linear Correlation')
elif((correlation1>=-0.01)&(correlation1<=-0.29)):
    print('Weak Negative Correlation')
elif((correlation1>=-0.30)& (correlation1<=-0.69)):
    print('Moderate Negative correlation')
elif((correlation1>=-0.70)& (correlation1<=-0.99)):
    print('Strong Negative correlation')
elif(correlation1==-1):
    print('Perfect Negative Correlation')
print('---------------------------------------------------')
# Q3: Do students with attendance of at least 80% have higher average scores than students below 80%?
attendance_above80=round(df[df['Attendance']>=80]['Final_Score'].mean(),2)
attendance_below80=round(df[df['Attendance']<80]['Final_Score'].mean(),2)
print('---------------------------------------------------')
print('Students With attendace Above 80: ',attendance_above80)
print('Student With attendance Below 80: ',attendance_below80)
print('---------------------------------------------------')
# checking who scores higher scores
if(attendance_above80>attendance_below80):
    print('Yes. Student with attendance of above 80 have higher average scores')
elif(attendance_above80<attendance_below80):
    print('No. Student with attendace of above 80 have lower average scores')
else:
    print('The average scores are equal')
print('-------------------------------------------------------------')
# Q4: Which students have attendance below 50%?
attendance_below50=df[df['Attendance']<50]
print(attendance_below50.to_string())

print('-------------------------------------------------------------------------------------------------------------')
# Q5: Are there students with high attendance but low final scores?
high_attendance_low_perfomance=df[(df['Attendance']>80) & (df['Final_Score']<50)]
print(high_attendance_low_perfomance.to_string())
print('-------------------------------------------------------------------------------------------------------------')
# Q6: Are there students with low attendance but high final scores?
low_attendance_high_Perfomance=df[(df['Attendance']<50) & (df['Final_Score']>60)]
print(low_attendance_high_Perfomance.to_string())
```

```python
# Visual to show the relationship between attendace and final score
fig, ax=plt.subplots(figsize=(12,6))

clean_df1=df.dropna(subset=['Attendance','Final_Score'])

# Plotting a scatter plot
clean_df1.plot(x='Attendance',
               y='Final_Score',
               color='green',
               alpha=0.7,
               ax=ax)
ax.set_title('Relationship Between Attendance and Final Score')
ax.set_xlabel('Attendance(%)')
ax.set_ylabel('Final Score')

# Plotting a trend line
slope, intercept=np.polyfit(clean_df1['Attendance'],
                            clean_df1['Final_Score'],
                            1)

# Creatting x axis plots
x_line=np.linspace(clean_df1['Attendance'].min(),
                   clean_df1['Final_Score'].max(),
                   100)

# Creating y axis points
y_line= slope*x_line+intercept

# plot the line
ax.plot(x_line,
        y_line,
        c='b',
        zorder=3,
        ls=':',
        label='Trend Line',
        lw=4)

ax.legend()
ax.grid(axis='y')
plt.savefig('../Visualizations/Relationship Between Attendance and Final Score.png',
            dpi=300,
            bbox_inches='tight')
plt.tight_layout()
plt.show()
```

```python
# CATEGORY D
# Assignment scores and examination results
# Q1: What is the average assignment score?
average_assignment_score=round(df['Assignment_Score'].mean(),2)
print('Average Assignment Score: ',average_assignment_score)

print('----------------------------------------------------------------------')
# Q2: What is the relationship between assignment scores and final examination scores?
correlation2=round(df['Assignment_Score'].corr(df['Final_Score']),6)
print('Relationship between Assignment Score and Final Score: ',correlation2)
# Identifying the strength of correlation
if(correlation2==1.00):
    print('Perfect Positive Correlation')
elif((correlation2>=0.70) & (correlation2<=0.99)):
    print('Strong Positive Correlation')
elif((correlation2>=0.30)& (correlation2<=0.68)):
    print('Moderate Positive Correlation')
elif((correlation2>=0.01)& (correlation2<=0.29)):
    print('Weak Positive correlation')
elif(correlation2==0.00):
    print('No Linear Correlation')
elif((correlation2>=-0.01)&(correlation2<=-0.29)):
    print('Weak Negative Correlation')
elif((correlation2>=-0.30)& (correlation2<=-0.69)):
    print('Moderate Negative correlation')
elif((correlation2>=-0.70)& (correlation2<=-0.99)):
    print('Strong Negative correlation')
elif(correlation2==-1):
    print('Perfect Negative Correlation')
print('----------------------------------------------------------------------')    
# Q3: Do students with high assignment scores generally achieve high final scores?
high_assignment_score=df[df['Assignment_Score']>75]['Final_Score'].mean()
low_assignment_score=df[df['Assignment_Score']<75]['Final_Score'].mean()

if(high_assignment_score>=low_assignment_score):
    print('Yes .High assigmnent scores students have high final scores')
elif(low_assignment_score>high_assignment_score):
    print('No.High assignment scores do not have high final scores')
print('----------------------------------------------------------------------')    
# Q4: Which students perform well in assignments but poorly in final examinations?
low_performers1_final_score=df[(df['Assignment_Score']>60) & (df['Final_Score']<50)]
print(low_performers1_final_score.to_string())
print('----------------------------------------------------------------------')
# Q5: Which students perform poorly in assignments but well in final examinations?
high_performers_final_score=df[(df['Assignment_Score']<50) & (df['Final_Score']>60)]
print(high_performers_final_score.to_string())
# Average final score among students with assignment scores above 75
avg_final_score_above_75=df[df['Assignment_Score']>75]['Final_Score'].mean().round(3)
print('Average Final Score above 75: ',avg_final_score_above_75)
    
```

```python
# Category E: Differences between student groups
# Q1: How many students belong to each gender category?
# Gender column cleaning
# Removing leading and trailing spaces from the gender column
df['Gender']=df['Gender'].str.strip()

# Converting all values to lower case
df['Gender']=df['Gender'].str.lower()

# Replacing inconsistent values in gender column
df['Gender']=df['Gender'].replace({
    'f':'female',
    'm':'male'
})

# Capitalize the standadized category
df['Gender']=df['Gender'].str.capitalize()

# Count the number of students in each gender category
gender_counts=df['Gender'].value_counts()
print(gender_counts)
print('=========================================================================================')
# Q2: What is the average final score for each gender category?
average_final_score_by_gender=df.groupby('Gender')['Final_Score'].mean().round(2)
print(average_final_score_by_gender)
print('=========================================================================================')
# Q3: What is the median final score for each gender category?
median_final_score_by_gender=df.groupby('Gender')['Final_Score'].median().round(2) 
print(median_final_score_by_gender)
print('=========================================================================================')
# Q4: Which age group has the highest average score?
age_group_avg_score=df.groupby('Age')['Final_Score'].mean().round(2)
highest_avg_score=age_group_avg_score.max()
age_with_highest_score=age_group_avg_score.idxmax()
print(age_group_avg_score)
print('Age with the highest average score: ',age_with_highest_score)
print('Highest average score: ',highest_avg_score)
print('=========================================================================================')
# Q5: Which class has the highest average score?
# Standardizing class column
df['Class']=df['Class'].str.strip()
df['Class']=df['Class'].str.lower()
df['Class']=df['Class'].str.capitalize()
class_counts=df['Class'].value_counts()
print(class_counts)
print('=========================================================================================')
# finding the highest class 
class_avg_score=df.groupby('Class')['Final_Score'].mean().round(2)
print(class_avg_score)
print('=========================================================================================')
highest_class_avg_score=class_avg_score.idxmax()
highest_avg_score=class_avg_score.max()
print(highest_class_avg_score)
print(highest_avg_score)


```
